🔹 Módulo 1: Introdução ao Python
Neste módulo, vamos entender o que é Python, como instalá-lo e escrever nosso primeiro código.

1️⃣ O que é Python?
Python é uma linguagem de programação de alto nível, interpretada e fácil de aprender. Ela é muito usada em diversas áreas, como:

✅ Desenvolvimento Web
✅ Análise de Dados
✅ Inteligência Artificial
✅ Automação de Tarefas
✅ Desenvolvimento de Jogos

Uma das vantagens do Python é sua sintaxe simples e intuitiva, que faz o código parecer quase uma linguagem humana.

Exemplo de código Python:



```python
print("Olá, Mundo!")
```

2️⃣ Instalando o Python
Passo 1: Baixar e instalar
1️⃣ Acesse o site oficial: https://www.python.org/downloads/
2️⃣ Baixe a versão mais recente do Python para seu sistema (Windows, Mac ou Linux).
3️⃣ Durante a instalação, marque a opção "Add Python to PATH" (Adicionar Python ao PATH) – Isso facilita executar Python pelo terminal.

Passo 2: Verificar a instalação
Depois de instalar, abra o Prompt de Comando (Windows) ou Terminal (Mac/Linux) e digite:

```python
python --version
```

Se aparecer algo como Python 3.x.x, significa que foi instalado com sucesso! 🎉



3️⃣ Executando seu primeiro código Python
Usando o Terminal
1️⃣ Abra o Prompt de Comando (Windows) ou Terminal (Mac/Linux).
2️⃣ Digite python e pressione Enter. Isso abrirá o interpretador interativo.
3️⃣ Agora, digite:

```python
print("Olá, Mundo!")
```

4️⃣ Pressione Enter e veja a mágica acontecer! ✨

Criando um arquivo Python
1️⃣ Abra um editor de texto ou IDE (pode ser o VS Code, PyCharm ou até o Bloco de Notas).
2️⃣ Escreva o seguinte código:

```python
print("Meu primeiro script em Python!")
```

3️⃣ Salve o arquivo com a extensão .py, por exemplo: meu_script.py.
4️⃣ Para rodar o código, abra o terminal na pasta onde o arquivo foi salvo e execute

```python
python meu_script.py
```

Se tudo estiver certo, a mensagem "Meu primeiro script em Python!" será exibida na tela. 🎉

4️⃣ Comentários e boas práticas
Comentários são partes do código que não são executadas. Eles servem para explicar o que cada parte do código faz. Em Python, usamos # para comentários de uma linha:

```python
# Isso é um comentário
print("Isso será impresso!")  # Outro comentário
```

Para comentários de múltiplas linhas, usamos três aspas:

```python
"""
Este é um comentário
de múltiplas linhas.
"""
```

Boas práticas iniciais
✅ Sempre nomeie variáveis de forma clara (nome em vez de n)
✅ Siga a indentação (Python usa espaços para definir blocos de código)
✅ Use comentários para explicar trechos importantes
